```python
import pandas as pd

df = pd.read_csv("C:/Titanic-Dataset.csv")

df.head()
```

```python
print("DATASET INFORMATION")
df.info()

print("\nMISSING VALUES")
print(df.isnull().sum())

print("\nDUPLICATE ROWS")
print(df.duplicated().sum())
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())

print("Missing Age values:", df["Age"].isnull().sum())
```

```python
# STEP 1: Handle missing values in Age

age_missing_before = df["Age"].isnull().sum()
age_median = df["Age"].median()

df["Age"] = df["Age"].fillna(age_median)

age_missing_after = df["Age"].isnull().sum()

print("Age missing values before cleaning :", age_missing_before)
print("Median age used for imputation     :", age_median)
print("Age missing values after cleaning  :", age_missing_after)
```

```python
# STEP 2: Handle missing values in Cabin

cabin_missing_before = df["Cabin"].isnull().sum()

df["Cabin"] = df["Cabin"].fillna("Unknown")

cabin_missing_after = df["Cabin"].isnull().sum()

print("Cabin missing values before cleaning :", cabin_missing_before)
print("Cabin missing values after cleaning  :", cabin_missing_after)
```

```python
# STEP 3: Handle missing values in Embarked

embarked_missing_before = df["Embarked"].isnull().sum()
embarked_mode = df["Embarked"].mode()[0]

df["Embarked"] = df["Embarked"].fillna(embarked_mode)

embarked_missing_after = df["Embarked"].isnull().sum()

print("Embarked missing values before cleaning :", embarked_missing_before)
print("Mode used for imputation               :", embarked_mode)
print("Embarked missing values after cleaning  :", embarked_missing_after)
```

```python
# STEP 4: Check and remove duplicate rows

duplicates_before = df.duplicated().sum()

if duplicates_before > 0:
    df = df.drop_duplicates().reset_index(drop=True)

duplicates_after = df.duplicated().sum()

print("Duplicate rows before cleaning :", duplicates_before)
print("Duplicate rows after cleaning  :", duplicates_after)


# STEP 5: Standardize text columns

text_columns = df.select_dtypes(include="object").columns

for column in text_columns:
    df[column] = df[column].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )

print("\nText columns standardized:")
print(list(text_columns))
```

```python
# STEP 6: Validate and standardize data types

print("DATA TYPES BEFORE VALIDATION")
print(df.dtypes)

# Convert numerical columns to appropriate numeric types
numeric_columns = ["PassengerId", "Survived", "Pclass", "Age", "SibSp", "Parch", "Fare"]

for column in numeric_columns:
    df[column] = pd.to_numeric(df[column], errors="coerce")

# Convert categorical columns to string type
categorical_columns = ["Sex", "Cabin", "Embarked"]

for column in categorical_columns:
    df[column] = df[column].astype("string").str.strip()

print("\nDATA TYPES AFTER VALIDATION")
print(df.dtypes)
```

```python
# STEP 7: Final Data Quality Validation

print("=" * 50)
print("FINAL DATA QUALITY CHECK")
print("=" * 50)

print("\n1. Dataset Shape:")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\n2. Missing Values:")
print(df.isnull().sum())

print("\n3. Total Missing Values:")
print(df.isnull().sum().sum())

print("\n4. Duplicate Rows:")
print(df.duplicated().sum())

print("\n5. Data Types:")
print(df.dtypes)

print("\n" + "=" * 50)
print("VALIDATION COMPLETED")
print("=" * 50)
```

```python
# STEP 8: Export Cleaned Dataset

output_file = "Titanic-Cleaned.csv"

df.to_csv(output_file, index=False)

print("Cleaned dataset exported successfully!")
print("File name:", output_file)
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```
